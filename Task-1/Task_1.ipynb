```python
import pandas as pd
df = pd.read_csv("Sample - Superstore.csv", encoding="latin1")

print(df.shape)
```

```python
df.info()
```

```python
print(df.isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df["Product Name"] = df["Product Name"].str.strip()
```

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")
```

```python
df["Ship Date"] = pd.to_datetime(df["Ship Date"], errors="coerce")
```

```python
print(df[["Order Date", "Ship Date"]].head())
```

```python
print("Invalid Order Dates:", df["Order Date"].isnull().sum())
print("Invalid Ship Dates:", df["Ship Date"].isnull().sum())
```

```python
df["Postal Code"] = df["Postal Code"].astype(str).str.zfill(5)
```

```python
print(df["Postal Code"].head(20))
```

```python
print(df[["Sales", "Quantity", "Discount", "Profit"]].describe())
```

```python
print("Negative Sales:", (df["Sales"] < 0).sum())
print("Negative Quantity:", (df["Quantity"] < 0).sum())
print("Negative Profit:", (df["Profit"] < 0).sum())
```

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nMissing values:")
print(df.isnull().sum().sum())

print("\nDuplicate rows:")
print(df.duplicated().sum())
```

```python
print(
    (df["Product Name"] != df["Product Name"].str.strip()).sum()
)
```

```python
df.to_csv("cleaned_superstore.csv", index=False)
```
